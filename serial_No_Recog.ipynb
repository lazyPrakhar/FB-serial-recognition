```python
import os

img_path = "/content/Data" #to actual images folder path

# Get all files in the folder and sort them
files = sorted(os.listdir(img_path))

# Rename files numerically
for i, filename in enumerate(files, start=1):
    old_path = os.path.join(img_path, filename)
    new_path = os.path.join(img_path, f"{i}{os.path.splitext(filename)[1]}")
    os.rename(old_path, new_path)
```

### Img Preprocessing
Initially some images are unclear, have objects in the image and are not in the same orientation for detecting serial no.

```python
import cv2
```

```python

img_path = "/content/Data"

# Get all image files in the folder
image_files = [f for f in os.listdir(img_path)]


for filename in image_files:
    img_file = os.path.join(img_path, filename)
    img = cv2.imread(img_file)

    if img is not None:
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) #grayscale
        blurred = cv2.GaussianBlur(gray, (5, 5), 0) #gaussin blur
        edges = cv2.Canny(blurred, 50, 150) #canny edge detection

        preprocessed_img_path = os.path.join("/content/Preprocessed_Data", filename) #create a folder named preprocessed_Data
        os.makedirs(os.path.dirname(preprocessed_img_path), exist_ok=True)
        cv2.imwrite(preprocessed_img_path, edges)
    else:
        print(f"Error reading image: {img_file}")
```
